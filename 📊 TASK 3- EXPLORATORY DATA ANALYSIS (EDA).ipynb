## 🎯 Goal:
Understand relationship between study hours and student scores using statistics and visualization.

```python
import pandas as pd

df = pd.read_csv("https://raw.githubusercontent.com/AdiPersonalWorks/Random/master/student_scores%20-%20student_scores.csv")
df.describe()
```

## 📌 Statistical Insights:

- Shows average study hours and scores
- Helps understand data distribution
- Gives minimum and maximum values

```python
df.corr(numeric_only=True)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.set_style("darkgrid")

plt.rcParams.update({
    "axes.facecolor": "#0b1220",
    "figure.facecolor": "#0b1220",
    "axes.edgecolor": "#94a3b8",
    "axes.labelcolor": "#e2e8f0",
    "xtick.color": "#e2e8f0",
    "ytick.color": "#e2e8f0",
    "text.color": "#e2e8f0"
})
```

## 🎨 Visualization Theme Setup

- Applied dark theme for better visual appeal
- Using neon-style color palette
- Making graphs portfolio-ready and modern

```python
import seaborn as sns
import matplotlib.pyplot as plt

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="viridis")
plt.title("Correlation Heatmap")
plt.show()
```

## 🔥 Correlation Heatmap

- Shows relationship between study hours and scores
- Strong positive correlation observed
- Confirms suitability for linear regression

```python
plt.figure(figsize=(8,5))
sns.scatterplot(
    x=df['Hours'],
    y=df['Scores'],
    color="#00f5d4",
    s=100
)
plt.title("Study Hours vs Scores")
plt.show()
```

## 📈 Study Hours vs Scores

- Strong linear relationship observed
- More study hours → higher scores
- Ideal for regression modeling

```python
plt.figure(figsize=(8,5))
sns.histplot(df['Scores'], kde=True, color="#a855f7")
plt.title("Score Distribution")
plt.show()
```

## 📊 Score Distribution

- Most students fall in mid-score range
- Few students achieve very high scores
- Data is slightly concentrated in middle

```python
sns.regplot(
    x=df['Hours'],
    y=df['Scores'],
    color="#f43f5e",
    line_kws={"color": "#38bdf8"}
)

plt.title("Regression Line: Hours vs Scores")
plt.show()
```

## 📉 Regression Insight

- Linear trend confirmed
- Model can easily predict scores from study hours
- Strong positive correlation

```python
df_sorted = df.sort_values("Hours")

plt.figure(figsize=(10,5))
plt.bar(df_sorted["Hours"], df_sorted["Scores"], color="#22c55e")
plt.title("Hours vs Scores (Bar View)")
plt.show()
```

## 📊 Comparative View

- Shows clear improvement in scores with increased study hours
- Helps visualize pattern more clearly
